````python
#### Tutorial Code
####PyTorch has two primitives to work with data: torch.utils.data.DataLoader and torch.utils.data.Dataset.
#####Dataset stores samples and their corresponding labels, and DataLoader wraps an iterable around the Dataset.
import torch
from torch import nn
from torch.utils.data import DataLoader
from torchvision import datasets
from torchvision.transforms import ToTensor, Lambda, Compose
import matplotlib.pyplot as plt

# Download training data from open datasets.
##Every TorchVision Dataset includes two arguments:
##transform and target_transform to modify the samples and labels respectively.

training_data = datasets.FashionMNIST(
    root="data",
    train=True,
    download=True,
    transform=ToTensor(),
)

# Download test data from open datasets.
test_data = datasets.FashionMNIST(
    root="data",
    train=False,
    download=True,
    transform=ToTensor(),
)
````

We pass the Dataset as an argument to [DataLoader](https://docs.pytorch.org/tutorials/beginner/basics/data_tutorial.html). This conveniently wraps an iterable over our dataset and supports automatic batching, sampling, shuffling, and multiprocess data loading. Here we define a batch size of 64, i.e. each element in the dataloader iterable will return a batch of 64 features and labels.

````python
batch_size = 64

# Create data loaders.
train_dataloader = DataLoader(training_data, batch_size=batch_size)
test_dataloader = DataLoader(test_data, batch_size=batch_size)

for X, y in test_dataloader:
    print("Shape of X [N, C, H, W]: ", X.shape)
    print("Shape of y: ", y.shape, y.dtype)
    break
````

To define a neural network in PyTorch, we create a class that inherits from [nn.Module](https://docs.pytorch.org/docs/stable/generated/torch.nn.Module.html). We define the layers of the network in the init function and specify how data will pass through the network in the forward function. To accelerate operations in the neural network, we move it to the GPU if available.

````python
# Get cpu or gpu device for training.
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using {} device".format(device))

import torch.nn.functional as F

# Define model
class NeuralNetwork(nn.Module):
    def __init__(self):
        super(NeuralNetwork, self).__init__()
        self.flatten = nn.Flatten()
        self.linear_relu_stack = nn.Sequential(
            nn.Linear(28*28, 512),
            nn.ReLU(),
            nn.Linear(512, 512),
            nn.ReLU(),
            nn.Linear(512, 10)
       )

    def forward(self, x):
        x = self.flatten(x)
        logits = self.linear_relu_stack(x)

        return logits

model = NeuralNetwork().to(device)
print(model)
````

````python
###Define the loss function and the optimizer
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.SGD(model.parameters(), lr=1e-3)
````

In a single training loop, the model makes predictions on the training dataset (fed to it in batches), and backpropagates the prediction error to adjust the model’s parameters.

````python
def train(dataloader, model, loss_fn, optimizer):
    size = len(dataloader.dataset)
    model.train()
    for batch, (X, y) in enumerate(dataloader):
        X, y = X.to(device), y.to(device)

        # Compute prediction error
        pred = model(X)
        loss = loss_fn(pred, y)

        # Backpropagation
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        if batch % 100 == 0:
            loss, current = loss.item(), batch * len(X)
            print(f"loss: {loss:>7f}  [{current:>5d}/{size:>5d}]")
````

````python
##Define a test function
def test(dataloader, model, loss_fn):
    size = len(dataloader.dataset)
    num_batches = len(dataloader)
    model.eval()
    test_loss, correct = 0, 0
    with torch.no_grad():
        for X, y in dataloader:
            X, y = X.to(device), y.to(device)
            pred = model(X)
            test_loss += loss_fn(pred, y).item()
            correct += (pred.argmax(1) == y).type(torch.float).sum().item()
    test_loss /= num_batches
    correct /= size
    print(f"Test Error: \n Accuracy: {(100*correct):>0.1f}%, Avg loss: {test_loss:>8f} \n")
````

````python
#Train and test the model
epochs = 5
for t in range(epochs):
    print(f"Epoch {t+1}\n-------------------------------")
    train(train_dataloader, model, loss_fn, optimizer)
    test(test_dataloader, model, loss_fn)
print("Done!")
````

````python
# Now we will apply xavier initialization to the model
def init_weights(m):
    if isinstance(m, nn.Linear):
        nn.init.xavier_uniform_(m.weight)
        if m.bias is not None:
            nn.init.zeros_(m.bias)
````

````python
# Lets apply the xavier initialization to the model
model.apply(init_weights)

# Now we will train and test the model again with the xavier initialization
epochs = 5
for t in range(epochs):
    print(f"Epoch {t+1}\n-------------------------------")
    train(train_dataloader, model, loss_fn, optimizer)
    test(test_dataloader, model, loss_fn)
print("Done!")
````

````python
# Modified training and testing functions to track losses and accuracies
def train_with_tracking(dataloader, model, loss_fn, optimizer):
    size = len(dataloader.dataset)
    model.train()
    total_loss = 0
    num_batches = len(dataloader)
    
    for batch, (X, y) in enumerate(dataloader):
        X, y = X.to(device), y.to(device)
        
        # Compute prediction error
        pred = model(X)
        loss = loss_fn(pred, y)
        total_loss += loss.item()
        
        # Backpropagation
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        if batch % 100 == 0:
            loss_val, current = loss.item(), batch * len(X)
            print(f"loss: {loss_val:>7f}  [{current:>5d}/{size:>5d}]")
    
    return total_loss / num_batches

def test_with_tracking(dataloader, model, loss_fn):
    size = len(dataloader.dataset)
    num_batches = len(dataloader)
    model.eval()
    test_loss, correct = 0, 0
    
    with torch.no_grad():
        for X, y in dataloader:
            X, y = X.to(device), y.to(device)
            pred = model(X)
            test_loss += loss_fn(pred, y).item()
            correct += (pred.argmax(1) == y).type(torch.float).sum().item()
    
    test_loss /= num_batches
    accuracy = correct / size
    print(f"Test Error: \n Accuracy: {(100*accuracy):>0.1f}%, Avg loss: {test_loss:>8f} \n")
    
    return test_loss, accuracy

# Training with different learning rates for 10 epochs
learning_rates = [1, 0.1, 0.01, 0.001]
epochs = 10
results_q12 = {}

plt.figure(figsize=(15, 10))

for lr_idx, lr in enumerate(learning_rates):
    print(f"\n{'='*50}")
    print(f"Training with Learning Rate: {lr}")
    print(f"{'='*50}")
    
    # Create a fresh model for each learning rate
    model = NeuralNetwork().to(device)
    loss_fn = nn.CrossEntropyLoss()
    optimizer = torch.optim.SGD(model.parameters(), lr=lr)
    
    train_losses = []
    test_losses = []
    test_accuracies = []
    
    for t in range(epochs):
        print(f"Epoch {t+1}\n-------------------------------")
        train_loss = train_with_tracking(train_dataloader, model, loss_fn, optimizer)
        test_loss, test_accuracy = test_with_tracking(test_dataloader, model, loss_fn)
        
        train_losses.append(train_loss)
        test_losses.append(test_loss)
        test_accuracies.append(test_accuracy)
    
    # Store results
    results_q12[lr] = {
        'final_accuracy': test_accuracies[-1],
        'train_losses': train_losses,
        'test_losses': test_losses,
        'test_accuracies': test_accuracies
    }
    
    # Plot training loss
    plt.subplot(2, 2, 1)
    plt.plot(range(1, epochs+1), train_losses, label=f'LR={lr}', marker='o')
    plt.title('Training Loss Curves')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    
    # Plot test loss
    plt.subplot(2, 2, 2)
    plt.plot(range(1, epochs+1), test_losses, label=f'LR={lr}', marker='s')
    plt.title('Test Loss Curves')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    
    # Plot test accuracy
    plt.subplot(2, 2, 3)
    plt.plot(range(1, epochs+1), [acc*100 for acc in test_accuracies], label=f'LR={lr}', marker='^')
    plt.title('Test Accuracy Curves')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy (%)')
    plt.legend()
    plt.grid(True)
    
    # Plot both training and test loss for comparison
    plt.subplot(2, 2, 4)
    plt.plot(range(1, epochs+1), train_losses, '--', label=f'Train LR={lr}', alpha=0.7)
    plt.plot(range(1, epochs+1), test_losses, '-', label=f'Test LR={lr}')

plt.subplot(2, 2, 4)
plt.title('Training vs Test Loss Comparison')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# Display results in table format
print("\nTable 1: Accuracy comparison for different learning rates (10 epochs)")
print("|Lr    |Accuracy|")
print("|------|--------|")
for lr in learning_rates:
    accuracy = results_q12[lr]['final_accuracy'] * 100
    print(f"|{lr:<6}|{accuracy:>7.2f}%|")
````

## Q1.3: Convergence Analysis with Extended Training (6 points)

For this analysis, I will define convergence criteria as:
1. **Primary criterion**: When the test accuracy improvement over the last 5 epochs is less than 0.1%
2. **Secondary criterion**: When the test loss doesn't decrease for 5 consecutive epochs
3. **Maximum epochs**: 50 epochs to ensure comprehensive analysis

This approach balances computational efficiency with thorough convergence detection.


````python
# Extended training for convergence analysis
def check_convergence(accuracies, losses, window=5, acc_threshold=0.001, loss_patience=5):
    """
    Check if model has converged based on accuracy improvement and loss stagnation
    """
    if len(accuracies) < window:
        return False, "Not enough epochs"
    
    # Check accuracy improvement in last 'window' epochs
    recent_accs = accuracies[-window:]
    acc_improvement = max(recent_accs) - min(recent_accs)
    
    # Check loss stagnation
    if len(losses) >= loss_patience:
        recent_losses = losses[-loss_patience:]
        loss_increasing = all(recent_losses[i] >= recent_losses[i-1] for i in range(1, len(recent_losses)))
    else:
        loss_increasing = False
    
    if acc_improvement < acc_threshold:
        return True, f"Accuracy converged (improvement < {acc_threshold*100:.1f}%)"
    elif loss_increasing:
        return True, "Loss stopped decreasing"
    
    return False, "Still training"

# Extended training with convergence detection
learning_rates = [1, 0.1, 0.01, 0.001]
max_epochs = 50
results_q13 = {}

plt.figure(figsize=(20, 12))

for lr_idx, lr in enumerate(learning_rates):
    print(f"\n{'='*60}")
    print(f"Extended Training with Learning Rate: {lr}")
    print(f"{'='*60}")
    
    # Create a fresh model for each learning rate
    model = NeuralNetwork().to(device)
    loss_fn = nn.CrossEntropyLoss()
    optimizer = torch.optim.SGD(model.parameters(), lr=lr)
    
    train_losses = []
    test_losses = []
    test_accuracies = []
    converged = False
    convergence_epoch = max_epochs
    convergence_reason = "Maximum epochs reached"
    
    for t in range(max_epochs):
        print(f"Epoch {t+1}\n-------------------------------")
        train_loss = train_with_tracking(train_dataloader, model, loss_fn, optimizer)
        test_loss, test_accuracy = test_with_tracking(test_dataloader, model, loss_fn)
        
        train_losses.append(train_loss)
        test_losses.append(test_loss)
        test_accuracies.append(test_accuracy)
        
        # Check for convergence
        if not converged:
            conv_status, conv_reason = check_convergence(test_accuracies, test_losses)
            if conv_status:
                converged = True
                convergence_epoch = t + 1
                convergence_reason = conv_reason
                print(f"*** CONVERGENCE DETECTED at epoch {convergence_epoch}: {convergence_reason} ***")
    
    # Find best accuracy and its epoch
    best_accuracy = max(test_accuracies)
    best_epoch = test_accuracies.index(best_accuracy) + 1
    
    # Store results
    results_q13[lr] = {
        'best_accuracy': best_accuracy,
        'best_epoch': best_epoch,
        'convergence_epoch': convergence_epoch,
        'convergence_reason': convergence_reason,
        'final_accuracy': test_accuracies[-1],
        'train_losses': train_losses,
        'test_losses': test_losses,
        'test_accuracies': test_accuracies
    }
    
    # Plot training loss
    plt.subplot(3, 2, 1)
    plt.plot(range(1, len(train_losses)+1), train_losses, label=f'LR={lr}', marker='o', markersize=2)
    plt.axvline(x=convergence_epoch, color=plt.gca().lines[-1].get_color(), linestyle='--', alpha=0.5)
    plt.title('Training Loss Curves (Extended)')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    
    # Plot test loss
    plt.subplot(3, 2, 2)
    plt.plot(range(1, len(test_losses)+1), test_losses, label=f'LR={lr}', marker='s', markersize=2)
    plt.axvline(x=convergence_epoch, color=plt.gca().lines[-1].get_color(), linestyle='--', alpha=0.5)
    plt.title('Test Loss Curves (Extended)')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    
    # Plot test accuracy
    plt.subplot(3, 2, 3)
    plt.plot(range(1, len(test_accuracies)+1), [acc*100 for acc in test_accuracies], 
             label=f'LR={lr}', marker='^', markersize=2)
    plt.axvline(x=best_epoch, color=plt.gca().lines[-1].get_color(), linestyle=':', alpha=0.7, label=f'Best LR={lr}')
    plt.axvline(x=convergence_epoch, color=plt.gca().lines[-1].get_color(), linestyle='--', alpha=0.5)
    plt.title('Test Accuracy Curves (Extended)')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy (%)')
    plt.legend()
    plt.grid(True)
    
    # Plot training vs test loss (first 20 epochs for clarity)
    plt.subplot(3, 2, 4)
    epochs_to_show = min(20, len(train_losses))
    plt.plot(range(1, epochs_to_show+1), train_losses[:epochs_to_show], '--', 
             label=f'Train LR={lr}', alpha=0.7)
    plt.plot(range(1, epochs_to_show+1), test_losses[:epochs_to_show], '-', 
             label=f'Test LR={lr}')
    plt.title('Training vs Test Loss (First 20 epochs)')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    
    # Plot accuracy improvement rate
    plt.subplot(3, 2, 5)
    if len(test_accuracies) > 1:
        acc_improvements = [test_accuracies[i] - test_accuracies[i-1] for i in range(1, len(test_accuracies))]
        plt.plot(range(2, len(test_accuracies)+1), [imp*100 for imp in acc_improvements], 
                 label=f'LR={lr}', marker='d', markersize=2)
    plt.title('Accuracy Improvement per Epoch')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy Improvement (%)')
    plt.legend()
    plt.grid(True)
    
    # Plot loss smoothed (moving average)
    plt.subplot(3, 2, 6)
    window_size = 3
    if len(test_losses) >= window_size:
        smoothed_losses = []
        for i in range(window_size-1, len(test_losses)):
            smoothed_losses.append(np.mean(test_losses[i-window_size+1:i+1]))
        plt.plot(range(window_size, len(test_losses)+1), smoothed_losses, 
                 label=f'LR={lr}', marker='x', markersize=2)
    plt.title(f'Smoothed Test Loss (Window={window_size})')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)

plt.tight_layout()
plt.show()

# Display detailed results
print(f"\n{'='*80}")
print("CONVERGENCE ANALYSIS RESULTS")
print(f"{'='*80}")

print("\nTable 2: Accuracy comparison for different learning rates (when converged)")
print("|Lr    |Best Accuracy|Best Epoch|Convergence Epoch|Convergence Reason|")
print("|------|-------------|----------|-----------------|------------------|")

for lr in learning_rates:
    result = results_q13[lr]
    best_acc = result['best_accuracy'] * 100
    best_ep = result['best_epoch']
    conv_ep = result['convergence_epoch']
    conv_reason = result['convergence_reason']
    print(f"|{lr:<6}|{best_acc:>11.2f}%|{best_ep:>10}|{conv_ep:>17}|{conv_reason:<18}|")

print(f"\n{'='*80}")
print("DETAILED ANALYSIS:")
print(f"{'='*80}")

for lr in learning_rates:
    result = results_q13[lr]
    print(f"\nLearning Rate {lr}:")
    print(f"  • Best accuracy: {result['best_accuracy']*100:.2f}% at epoch {result['best_epoch']}")
    print(f"  • Final accuracy: {result['final_accuracy']*100:.2f}%")
    print(f"  • Convergence: Epoch {result['convergence_epoch']} ({result['convergence_reason']})")
    print(f"  • Training completed: {len(result['train_losses'])} epochs")
````

````python
# Gradient norm analysis
def compute_gradient_norm(model):
    """Compute the mean of gradient norms for all trainable parameters"""
    total_norm = 0.0
    param_count = 0
    
    for param in model.parameters():
        if param.grad is not None:
            param_norm = param.grad.data.norm(2)  # L2 norm
            total_norm += param_norm.item()
            param_count += 1
    
    return total_norm / param_count if param_count > 0 else 0.0

def train_with_gradient_tracking(dataloader, model, loss_fn, optimizer, sample_steps=None):
    """Modified training function that tracks gradient norms at specified steps"""
    size = len(dataloader.dataset)
    model.train()
    total_loss = 0
    num_batches = len(dataloader)
    gradient_norms = []
    sampled_steps = []
    current_step = 0
    
    for batch, (X, y) in enumerate(dataloader):
        X, y = X.to(device), y.to(device)
        
        # Compute prediction error
        pred = model(X)
        loss = loss_fn(pred, y)
        total_loss += loss.item()
        
        # Backpropagation
        optimizer.zero_grad()
        loss.backward()
        
        # Track gradient norm if this step should be sampled
        if sample_steps is not None and current_step in sample_steps:
            grad_norm = compute_gradient_norm(model)
            gradient_norms.append(grad_norm)
            sampled_steps.append(current_step)
        
        optimizer.step()
        current_step += 1
        
        if batch % 100 == 0:
            loss_val, current_batch = loss.item(), batch * len(X)
            print(f"loss: {loss_val:>7f}  [{current_batch:>5d}/{size:>5d}] step: {current_step}")
    
    return total_loss / num_batches, gradient_norms, sampled_steps

# Setup for gradient tracking experiment
print("Setting up gradient norm tracking experiment...")

# Use base model for this experiment
model = BaseNetwork().to(device)
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# Calculate total steps and sampling interval
epochs_for_gradients = 50
total_steps = len(train_dataloader) * epochs_for_gradients
sample_interval = max(1, total_steps // 100)  # Sample 100 steps
sample_steps = set(range(0, total_steps, sample_interval))

print(f"Total training steps: {total_steps}")
print(f"Sampling every {sample_interval} steps")
print(f"Total samples expected: {len(sample_steps)}")

# Training with gradient tracking
all_gradient_norms = []
all_sampled_steps = []
train_losses = []
test_losses = []
test_accuracies = []

print(f"\n{'='*60}")
print("Training with Gradient Norm Tracking")
print(f"{'='*60}")

for t in range(epochs_for_gradients):
    print(f"Epoch {t+1}\n-------------------------------")
    
    # Training with gradient tracking
    train_loss, epoch_grad_norms, epoch_sampled_steps = train_with_gradient_tracking(
        train_dataloader, model, loss_fn, optimizer, sample_steps
    )
    
    # Update global step counter for next epoch
    epoch_start_step = t * len(train_dataloader)
    epoch_sampled_steps_adjusted = [step - epoch_start_step for step in epoch_sampled_steps]
    
    all_gradient_norms.extend(epoch_grad_norms)
    all_sampled_steps.extend([step + epoch_start_step for step in epoch_sampled_steps_adjusted])
    
    # Test evaluation
    test_loss, test_accuracy = test_with_tracking(test_dataloader, model, loss_fn)
    
    train_losses.append(train_loss)
    test_losses.append(test_loss)
    test_accuracies.append(test_accuracy)
    
    # Remove processed steps from sample_steps for next epoch
    steps_to_remove = set(range(epoch_start_step, epoch_start_step + len(train_dataloader)))
    sample_steps = sample_steps - steps_to_remove

print(f"\nGradient tracking completed!")
print(f"Total samples collected: {len(all_gradient_norms)}")

# Create comprehensive plots
plt.figure(figsize=(20, 12))

# Plot 1: Gradient norm over training steps
plt.subplot(3, 3, 1)
plt.plot(all_sampled_steps, all_gradient_norms, 'b-', alpha=0.7, marker='o', markersize=2)
plt.title('Mean Gradient Norm Over Training Steps')
plt.xlabel('Training Step')
plt.ylabel('Mean Gradient Norm')
plt.grid(True)

# Plot 2: Gradient norm over epochs (averaged per epoch)
plt.subplot(3, 3, 2)
# Group gradient norms by epoch
steps_per_epoch = len(train_dataloader)
gradient_norms_by_epoch = []
epoch_numbers = []

for epoch in range(epochs_for_gradients):
    epoch_start = epoch * steps_per_epoch
    epoch_end = (epoch + 1) * steps_per_epoch
    
    # Find gradients for this epoch
    epoch_gradients = [norm for step, norm in zip(all_sampled_steps, all_gradient_norms) 
                       if epoch_start <= step < epoch_end]
    
    if epoch_gradients:
        gradient_norms_by_epoch.append(np.mean(epoch_gradients))
        epoch_numbers.append(epoch + 1)

plt.plot(epoch_numbers, gradient_norms_by_epoch, 'r-', marker='s', markersize=4)
plt.title('Mean Gradient Norm Per Epoch')
plt.xlabel('Epoch')
plt.ylabel('Mean Gradient Norm')
plt.grid(True)

# Plot 3: Training loss over epochs
plt.subplot(3, 3, 3)
plt.plot(range(1, len(train_losses)+1), train_losses, 'g-', marker='^', markersize=3)
plt.title('Training Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Training Loss')
plt.grid(True)

# Plot 4: Test accuracy over epochs
plt.subplot(3, 3, 4)
plt.plot(range(1, len(test_accuracies)+1), [acc*100 for acc in test_accuracies], 'purple', marker='d', markersize=3)
plt.title('Test Accuracy Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Test Accuracy (%)')
plt.grid(True)

# Plot 5: Gradient norm vs training loss correlation
plt.subplot(3, 3, 5)
# Interpolate gradient norms to match epoch-based loss measurements
if len(gradient_norms_by_epoch) == len(train_losses):
    plt.scatter(gradient_norms_by_epoch, train_losses, alpha=0.6, c=range(len(train_losses)), cmap='viridis')
    plt.xlabel('Mean Gradient Norm')
    plt.ylabel('Training Loss')
    plt.title('Gradient Norm vs Training Loss')
    plt.colorbar(label='Epoch')
    plt.grid(True)

# Plot 6: Gradient norm smoothed (moving average)
plt.subplot(3, 3, 6)
window_size = max(1, len(all_gradient_norms) // 20)  # 5% of data points
if len(all_gradient_norms) >= window_size:
    smoothed_gradients = []
    smoothed_steps = []
    for i in range(window_size-1, len(all_gradient_norms)):
        smoothed_gradients.append(np.mean(all_gradient_norms[i-window_size+1:i+1]))
        smoothed_steps.append(all_sampled_steps[i])
    
    plt.plot(smoothed_steps, smoothed_gradients, 'orange', linewidth=2)
    plt.plot(all_sampled_steps, all_gradient_norms, 'lightblue', alpha=0.3, linewidth=1)
    plt.title(f'Smoothed Gradient Norm (Window={window_size})')
    plt.xlabel('Training Step')
    plt.ylabel('Mean Gradient Norm')
    plt.legend(['Smoothed', 'Original'])
    plt.grid(True)

# Plot 7: Gradient norm distribution
plt.subplot(3, 3, 7)
plt.hist(all_gradient_norms, bins=30, alpha=0.7, color='skyblue', edgecolor='black')
plt.title('Distribution of Gradient Norms')
plt.xlabel('Mean Gradient Norm')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

# Plot 8: Log scale gradient norm
plt.subplot(3, 3, 8)
plt.plot(all_sampled_steps, all_gradient_norms, 'b-', alpha=0.7, marker='o', markersize=1)
plt.title('Mean Gradient Norm (Log Scale)')
plt.xlabel('Training Step')
plt.ylabel('Mean Gradient Norm (log)')
plt.yscale('log')
plt.grid(True)

# Plot 9: Gradient norm change rate
plt.subplot(3, 3, 9)
if len(all_gradient_norms) > 1:
    gradient_changes = [all_gradient_norms[i] - all_gradient_norms[i-1] 
                       for i in range(1, len(all_gradient_norms))]
    change_steps = all_sampled_steps[1:]
    plt.plot(change_steps, gradient_changes, 'red', alpha=0.7, marker='x', markersize=2)
    plt.title('Gradient Norm Change Rate')
    plt.xlabel('Training Step')
    plt.ylabel('Gradient Norm Change')
    plt.axhline(y=0, color='black', linestyle='--', alpha=0.5)
    plt.grid(True)

plt.tight_layout()
plt.show()

# Statistical analysis of gradient norms
print(f"\n{'='*80}")
print("GRADIENT NORM ANALYSIS RESULTS")
print(f"{'='*80}")

print(f"\nGradient Norm Statistics:")
print(f"• Total samples: {len(all_gradient_norms)}")
print(f"• Mean gradient norm: {np.mean(all_gradient_norms):.6f}")
print(f"• Std deviation: {np.std(all_gradient_norms):.6f}")
print(f"• Min gradient norm: {np.min(all_gradient_norms):.6f}")
print(f"• Max gradient norm: {np.max(all_gradient_norms):.6f}")
print(f"• Initial gradient norm (first sample): {all_gradient_norms[0]:.6f}")
print(f"• Final gradient norm (last sample): {all_gradient_norms[-1]:.6f}")

# Trend analysis
initial_avg = np.mean(all_gradient_norms[:len(all_gradient_norms)//4])  # First quarter
final_avg = np.mean(all_gradient_norms[-len(all_gradient_norms)//4:])  # Last quarter
trend = "decreasing" if final_avg < initial_avg else "increasing"
change_magnitude = abs(final_avg - initial_avg) / initial_avg * 100

print(f"\nTrend Analysis:")
print(f"• Initial average (first 25%): {initial_avg:.6f}")
print(f"• Final average (last 25%): {final_avg:.6f}")
print(f"• Overall trend: {trend}")
print(f"• Change magnitude: {change_magnitude:.2f}%")

print(f"\n{'='*80}")
print("INTERPRETATION OF GRADIENT NORM TRENDS:")
print(f"{'='*80}")

print(f"""
The gradient norm behavior observed during training provides insights into the optimization dynamics:

1. **Initial High Gradients**: Early in training, gradient norms are typically higher because:
   - The model is far from the optimal solution
   - Large prediction errors lead to large gradients
   - The loss landscape is steeper in the initial parameter space

2. **Gradual Decrease**: As training progresses, gradient norms generally decrease because:
   - The model approaches a local minimum where gradients are smaller
   - The loss becomes smoother in the vicinity of the solution
   - The model's predictions become more accurate, reducing error signals

3. **Convergence Indicator**: Very small gradient norms indicate:
   - The model is close to a local minimum
   - Learning rate might need adjustment if gradients become too small
   - Potential convergence of the optimization process

4. **Optimization Stability**: Consistent gradient norm patterns suggest:
   - Stable optimization without exploding or vanishing gradients
   - Appropriate learning rate selection
   - Healthy training dynamics

The observed trend of {trend} gradient norms by {change_magnitude:.1f}% indicates that the model is 
successfully converging towards a local minimum, with the optimization process behaving as expected 
for a well-tuned neural network training setup.
""")

# Correlation with training metrics
if len(gradient_norms_by_epoch) == len(train_losses):
    grad_loss_corr = np.corrcoef(gradient_norms_by_epoch, train_losses)[0, 1]
    print(f"\nCorrelation between gradient norm and training loss: {grad_loss_corr:.4f}")
    
    if abs(grad_loss_corr) > 0.7:
        print("Strong correlation indicates gradient norms are a good predictor of training progress.")
    elif abs(grad_loss_corr) > 0.4:
        print("Moderate correlation shows some relationship between gradient norms and loss.")
    else:
        print("Weak correlation suggests gradient norms and loss have independent dynamics.")
````

````python
# Gradient norm analysis - FIXED VERSION
def compute_gradient_norm(model):
    """Compute the mean of gradient norms for all trainable parameters"""
    total_norm = 0.0
    param_count = 0
    
    for param in model.parameters():
        if param.grad is not None:
            param_norm = param.grad.data.norm(2)  # L2 norm
            total_norm += param_norm.item()
            param_count += 1
    
    return total_norm / param_count if param_count > 0 else 0.0

def train_with_gradient_tracking(dataloader, model, loss_fn, optimizer, sample_frequency=50):
    """Modified training function that tracks gradient norms at regular intervals"""
    size = len(dataloader.dataset)
    model.train()
    total_loss = 0
    num_batches = len(dataloader)
    gradient_norms = []
    sampled_steps = []
    
    for batch, (X, y) in enumerate(dataloader):
        X, y = X.to(device), y.to(device)
        
        # Compute prediction error
        pred = model(X)
        loss = loss_fn(pred, y)
        total_loss += loss.item()
        
        # Backpropagation
        optimizer.zero_grad()
        loss.backward()
        
        # Track gradient norm at regular intervals
        if batch % sample_frequency == 0:
            grad_norm = compute_gradient_norm(model)
            gradient_norms.append(grad_norm)
            sampled_steps.append(batch)
        
        optimizer.step()
        
        if batch % 100 == 0:
            loss_val, current_batch = loss.item(), batch * len(X)
            print(f"loss: {loss_val:>7f}  [{current_batch:>5d}/{size:>5d}] batch: {batch}")
    
    return total_loss / num_batches, gradient_norms, sampled_steps

# Setup for gradient tracking experiment
print("Setting up gradient norm tracking experiment...")

# Use base model for this experiment
model = BaseNetwork().to(device)
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# Training configuration
epochs_for_gradients = 50
sample_frequency = 50  # Sample every 50 batches
batches_per_epoch = len(train_dataloader)

print(f"Total epochs: {epochs_for_gradients}")
print(f"Batches per epoch: {batches_per_epoch}")
print(f"Sampling every {sample_frequency} batches")
print(f"Expected samples per epoch: {batches_per_epoch // sample_frequency}")
print(f"Total expected samples: {epochs_for_gradients * (batches_per_epoch // sample_frequency)}")

# Training with gradient tracking
all_gradient_norms = []
all_sampled_steps = []
train_losses = []
test_losses = []
test_accuracies = []

print(f"\n{'='*60}")
print("Training with Gradient Norm Tracking")
print(f"{'='*60}")

for epoch in range(epochs_for_gradients):
    print(f"Epoch {epoch+1}\n-------------------------------")
    
    # Training with gradient tracking
    train_loss, epoch_grad_norms, epoch_sampled_steps = train_with_gradient_tracking(
        train_dataloader, model, loss_fn, optimizer, sample_frequency
    )
    
    # Convert epoch-relative steps to global steps
    global_steps = [step + epoch * batches_per_epoch for step in epoch_sampled_steps]
    
    all_gradient_norms.extend(epoch_grad_norms)
    all_sampled_steps.extend(global_steps)
    
    # Test evaluation
    test_loss, test_accuracy = test_with_tracking(test_dataloader, model, loss_fn)
    
    train_losses.append(train_loss)
    test_losses.append(test_loss)
    test_accuracies.append(test_accuracy)
    
    print(f"Epoch {epoch+1} completed. Gradient samples collected: {len(epoch_grad_norms)}")

print(f"\nGradient tracking completed!")
print(f"Total samples collected: {len(all_gradient_norms)}")

# Create comprehensive plots
plt.figure(figsize=(20, 12))

# Plot 1: Gradient norm over training steps
plt.subplot(3, 3, 1)
plt.plot(all_sampled_steps, all_gradient_norms, 'b-', alpha=0.7, marker='o', markersize=2)
plt.title('Mean Gradient Norm Over Training Steps')
plt.xlabel('Training Step')
plt.ylabel('Mean Gradient Norm')
plt.grid(True)

# Plot 2: Gradient norm over epochs (averaged per epoch)
plt.subplot(3, 3, 2)
# Group gradient norms by epoch
gradient_norms_by_epoch = []
epoch_numbers = []

for epoch in range(epochs_for_gradients):
    epoch_start = epoch * batches_per_epoch
    epoch_end = (epoch + 1) * batches_per_epoch
    
    # Find gradients for this epoch
    epoch_gradients = [norm for step, norm in zip(all_sampled_steps, all_gradient_norms) 
                       if epoch_start <= step < epoch_end]
    
    if epoch_gradients:
        gradient_norms_by_epoch.append(np.mean(epoch_gradients))
        epoch_numbers.append(epoch + 1)

plt.plot(epoch_numbers, gradient_norms_by_epoch, 'r-', marker='s', markersize=4)
plt.title('Mean Gradient Norm Per Epoch')
plt.xlabel('Epoch')
plt.ylabel('Mean Gradient Norm')
plt.grid(True)

# Plot 3: Training loss over epochs
plt.subplot(3, 3, 3)
plt.plot(range(1, len(train_losses)+1), train_losses, 'g-', marker='^', markersize=3)
plt.title('Training Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Training Loss')
plt.grid(True)

# Plot 4: Test accuracy over epochs
plt.subplot(3, 3, 4)
plt.plot(range(1, len(test_accuracies)+1), [acc*100 for acc in test_accuracies], 'purple', marker='d', markersize=3)
plt.title('Test Accuracy Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Test Accuracy (%)')
plt.grid(True)

# Plot 5: Gradient norm vs training loss correlation
plt.subplot(3, 3, 5)
if len(gradient_norms_by_epoch) == len(train_losses):
    plt.scatter(gradient_norms_by_epoch, train_losses, alpha=0.6, c=range(len(train_losses)), cmap='viridis')
    plt.xlabel('Mean Gradient Norm')
    plt.ylabel('Training Loss')
    plt.title('Gradient Norm vs Training Loss')
    plt.colorbar(label='Epoch')
    plt.grid(True)

# Plot 6: Gradient norm smoothed (moving average)
plt.subplot(3, 3, 6)
if len(all_gradient_norms) >= 5:
    window_size = max(5, len(all_gradient_norms) // 20)  # At least 5, or 5% of data points
    smoothed_gradients = []
    smoothed_steps = []
    for i in range(window_size-1, len(all_gradient_norms)):
        smoothed_gradients.append(np.mean(all_gradient_norms[i-window_size+1:i+1]))
        smoothed_steps.append(all_sampled_steps[i])
    
    plt.plot(smoothed_steps, smoothed_gradients, 'orange', linewidth=2)
    plt.plot(all_sampled_steps, all_gradient_norms, 'lightblue', alpha=0.3, linewidth=1)
    plt.title(f'Smoothed Gradient Norm (Window={window_size})')
    plt.xlabel('Training Step')
    plt.ylabel('Mean Gradient Norm')
    plt.legend(['Smoothed', 'Original'])
    plt.grid(True)

# Plot 7: Gradient norm distribution
plt.subplot(3, 3, 7)
plt.hist(all_gradient_norms, bins=min(30, len(all_gradient_norms)//2), alpha=0.7, color='skyblue', edgecolor='black')
plt.title('Distribution of Gradient Norms')
plt.xlabel('Mean Gradient Norm')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

# Plot 8: Log scale gradient norm
plt.subplot(3, 3, 8)
plt.semilogy(all_sampled_steps, all_gradient_norms, 'b-', alpha=0.7, marker='o', markersize=1)
plt.title('Mean Gradient Norm (Log Scale)')
plt.xlabel('Training Step')
plt.ylabel('Mean Gradient Norm (log)')
plt.grid(True)

# Plot 9: Gradient norm change rate
plt.subplot(3, 3, 9)
if len(all_gradient_norms) > 1:
    gradient_changes = [all_gradient_norms[i] - all_gradient_norms[i-1] 
                       for i in range(1, len(all_gradient_norms))]
    change_steps = all_sampled_steps[1:]
    plt.plot(change_steps, gradient_changes, 'red', alpha=0.7, marker='x', markersize=2)
    plt.title('Gradient Norm Change Rate')
    plt.xlabel('Training Step')
    plt.ylabel('Gradient Norm Change')
    plt.axhline(y=0, color='black', linestyle='--', alpha=0.5)
    plt.grid(True)

plt.tight_layout()
plt.show()

# Statistical analysis of gradient norms
print(f"\n{'='*80}")
print("GRADIENT NORM ANALYSIS RESULTS")
print(f"{'='*80}")

if len(all_gradient_norms) > 0:
    print(f"\nGradient Norm Statistics:")
    print(f"• Total samples: {len(all_gradient_norms)}")
    print(f"• Mean gradient norm: {np.mean(all_gradient_norms):.6f}")
    print(f"• Std deviation: {np.std(all_gradient_norms):.6f}")
    print(f"• Min gradient norm: {np.min(all_gradient_norms):.6f}")
    print(f"• Max gradient norm: {np.max(all_gradient_norms):.6f}")
    print(f"• Initial gradient norm (first sample): {all_gradient_norms[0]:.6f}")
    print(f"• Final gradient norm (last sample): {all_gradient_norms[-1]:.6f}")
    
    # Trend analysis
    if len(all_gradient_norms) >= 4:
        quarter_size = len(all_gradient_norms) // 4
        initial_avg = np.mean(all_gradient_norms[:quarter_size])  # First quarter
        final_avg = np.mean(all_gradient_norms[-quarter_size:])  # Last quarter
        trend = "decreasing" if final_avg < initial_avg else "increasing"
        change_magnitude = abs(final_avg - initial_avg) / initial_avg * 100
        
        print(f"\nTrend Analysis:")
        print(f"• Initial average (first 25%): {initial_avg:.6f}")
        print(f"• Final average (last 25%): {final_avg:.6f}")
        print(f"• Overall trend: {trend}")
        print(f"• Change magnitude: {change_magnitude:.2f}%")
        
        print(f"\n{'='*80}")
        print("INTERPRETATION OF GRADIENT NORM TRENDS:")
        print(f"{'='*80}")
        
        print(f"""
The gradient norm behavior observed during training provides insights into the optimization dynamics:

1. **Training Progression**: The gradient norms show a {trend} trend by {change_magnitude:.1f}%, indicating 
   {'the model is approaching convergence' if trend == 'decreasing' else 'ongoing optimization dynamics'}.

2. **Optimization Health**: Mean gradient norm of {np.mean(all_gradient_norms):.4f} suggests stable training
   without exploding or vanishing gradient problems.

3. **Convergence Indicator**: The {trend} trend indicates that the model is 
   {'successfully converging towards a local minimum' if trend == 'decreasing' else 'still actively learning'}.

4. **Stability Assessment**: Standard deviation of {np.std(all_gradient_norms):.4f} shows 
   {'consistent' if np.std(all_gradient_norms) < np.mean(all_gradient_norms) else 'variable'} gradient behavior.
        """)
        
        # Correlation with training metrics
        if len(gradient_norms_by_epoch) == len(train_losses) and len(gradient_norms_by_epoch) > 1:
            grad_loss_corr = np.corrcoef(gradient_norms_by_epoch, train_losses)[0, 1]
            print(f"\nCorrelation between gradient norm and training loss: {grad_loss_corr:.4f}")
            
            if abs(grad_loss_corr) > 0.7:
                print("Strong correlation indicates gradient norms are a good predictor of training progress.")
            elif abs(grad_loss_corr) > 0.4:
                print("Moderate correlation shows some relationship between gradient norms and loss.")
            else:
                print("Weak correlation suggests gradient norms and loss have independent dynamics.")
        else:
            print("\nNot enough data to compute reliable correlation with training loss.")
    else:
        print("\nInsufficient data to perform trend analysis.")
else:
    print("No gradient norm samples collected.")
    
print(f"\n{'='*80}")
print("Gradient norm tracking experiment concluded successfully.")
print(f"{'='*80}")
````

````python
# Define CNN architecture for Fashion-MNIST
class ConvolutionalNetwork(nn.Module):
    def __init__(self):
        super(ConvolutionalNetwork, self).__init__()
        
        # Convolutional layers
        self.conv_layers = nn.Sequential(
            # First conv block: 1 -> 32 channels
            nn.Conv2d(1, 32, kernel_size=3, padding=1),  # 28x28 -> 28x28
            nn.ReLU(),
            nn.Conv2d(32, 32, kernel_size=3, padding=1),  # 28x28 -> 28x28
            nn.ReLU(),
            nn.MaxPool2d(2, 2),  # 28x28 -> 14x14
            nn.Dropout(0.25),
            
            # Second conv block: 32 -> 64 channels
            nn.Conv2d(32, 64, kernel_size=3, padding=1),  # 14x14 -> 14x14
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),  # 14x14 -> 14x14
            nn.ReLU(),
            nn.MaxPool2d(2, 2),  # 14x14 -> 7x7
            nn.Dropout(0.25),
        )
        
        # Calculate the size after conv layers
        # After all conv layers: 64 channels * 7 * 7 = 3136
        
        # Fully connected layers
        self.fc_layers = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64 * 7 * 7, 128),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(128, 10)
        )
    
    def forward(self, x):
        x = self.conv_layers(x)
        x = self.fc_layers(x)
        return x

# Create a simpler CNN for fair comparison
class SimpleCNN(nn.Module):
    def __init__(self):
        super(SimpleCNN, self).__init__()
        
        self.conv_layers = nn.Sequential(
            # Single conv block
            nn.Conv2d(1, 16, kernel_size=5, padding=2),  # 28x28 -> 28x28
            nn.ReLU(),
            nn.MaxPool2d(2, 2),  # 28x28 -> 14x14
            
            nn.Conv2d(16, 32, kernel_size=5, padding=2),  # 14x14 -> 14x14
            nn.ReLU(),
            nn.MaxPool2d(2, 2),  # 14x14 -> 7x7
        )
        
        # Fully connected layers
        self.fc_layers = nn.Sequential(
            nn.Flatten(),
            nn.Linear(32 * 7 * 7, 128),
            nn.ReLU(),
            nn.Linear(128, 10)
        )
    
    def forward(self, x):
        x = self.conv_layers(x)
        x = self.fc_layers(x)
        return x

# Test both CNN architectures and compare with MLP
cnn_models = {
    'Simple_CNN': SimpleCNN(),
    'Full_CNN': ConvolutionalNetwork()
}

# Add MLP for comparison
mlp_model = {'MLP_Base': BaseNetwork()}

# Combine all models for comparison
all_models = {**cnn_models, **mlp_model}

# Print parameter counts
print("Model Parameter Comparison:")
print("="*50)
for name, model in all_models.items():
    param_count = count_parameters(model)
    print(f"{name:<12}: {param_count:>8,} parameters ({param_count/1_000_000:.2f}M)")

# Training configuration
fixed_lr = 0.01
epochs = 30
results_q17 = {}

plt.figure(figsize=(20, 15))

for model_idx, (model_name, model) in enumerate(all_models.items()):
    print(f"\n{'='*60}")
    print(f"Training {model_name}")
    print(f"{'='*60}")
    
    # Move model to device and set up training
    model = model.to(device)
    loss_fn = nn.CrossEntropyLoss()
    optimizer = torch.optim.SGD(model.parameters(), lr=fixed_lr)
    
    train_losses = []
    test_losses = []
    test_accuracies = []
    
    # Training loop
    for t in range(epochs):
        print(f"Epoch {t+1}\n-------------------------------")
        
        train_loss = train_with_tracking(train_dataloader, model, loss_fn, optimizer)
        test_loss, test_accuracy = test_with_tracking(test_dataloader, model, loss_fn)
        
        train_losses.append(train_loss)
        test_losses.append(test_loss)
        test_accuracies.append(test_accuracy)
    
    # Store results
    results_q17[model_name] = {
        'parameters': count_parameters(model),
        'final_accuracy': test_accuracies[-1],
        'best_accuracy': max(test_accuracies),
        'best_epoch': test_accuracies.index(max(test_accuracies)) + 1,
        'train_losses': train_losses,
        'test_losses': test_losses,
        'test_accuracies': test_accuracies,
        'model_type': 'CNN' if 'CNN' in model_name else 'MLP'
    }
    
    # Colors for different model types
    colors = {'Simple_CNN': 'blue', 'Full_CNN': 'red', 'MLP_Base': 'green'}
    color = colors.get(model_name, 'black')
    
    # Plot 1: Training Loss Comparison
    plt.subplot(3, 4, 1)
    plt.plot(range(1, epochs+1), train_losses, label=model_name, color=color, marker='o', markersize=2)
    plt.title('Training Loss: CNN vs MLP')
    plt.xlabel('Epoch')
    plt.ylabel('Training Loss')
    plt.legend()
    plt.grid(True)
    
    # Plot 2: Test Loss Comparison
    plt.subplot(3, 4, 2)
    plt.plot(range(1, epochs+1), test_losses, label=model_name, color=color, marker='s', markersize=2)
    plt.title('Test Loss: CNN vs MLP')
    plt.xlabel('Epoch')
    plt.ylabel('Test Loss')
    plt.legend()
    plt.grid(True)
    
    # Plot 3: Test Accuracy Comparison
    plt.subplot(3, 4, 3)
    plt.plot(range(1, epochs+1), [acc*100 for acc in test_accuracies], 
             label=model_name, color=color, marker='^', markersize=2)
    plt.title('Test Accuracy: CNN vs MLP')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy (%)')
    plt.legend()
    plt.grid(True)
    
    # Plot 4: Parameters vs Final Accuracy
    plt.subplot(3, 4, 4)
    plt.scatter(results_q17[model_name]['parameters'], 
               results_q17[model_name]['final_accuracy']*100, 
               s=100, label=model_name, color=color, alpha=0.7)
    plt.xlabel('Number of Parameters')
    plt.ylabel('Final Accuracy (%)')
    plt.title('Parameters vs Accuracy')
    plt.legend()
    plt.grid(True)

# Individual model analysis plots
model_names = list(results_q17.keys())
for i, model_name in enumerate(model_names):
    result = results_q17[model_name]
    
    # Training vs Test Loss for individual models
    plt.subplot(3, 4, 5 + i)
    plt.plot(range(1, epochs+1), result['train_losses'], '--', 
             label='Training', alpha=0.7, color=colors.get(model_name, 'black'))
    plt.plot(range(1, epochs+1), result['test_losses'], '-', 
             label='Test', color=colors.get(model_name, 'black'))
    plt.title(f'{model_name}: Train vs Test Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    
    # Accuracy progression
    plt.subplot(3, 4, 8 + i)
    plt.plot(range(1, epochs+1), [acc*100 for acc in result['test_accuracies']], 
             color=colors.get(model_name, 'black'), marker='o', markersize=2)
    plt.axhline(y=result['best_accuracy']*100, color='red', linestyle='--', alpha=0.5)
    plt.title(f'{model_name}: Accuracy Progression')
    plt.xlabel('Epoch')
    plt.ylabel('Test Accuracy (%)')
    plt.grid(True)

# Final comparison plot
plt.subplot(3, 4, 12)
model_types = []
accuracies = []
param_counts = []

for model_name, result in results_q17.items():
    model_types.append(model_name)
    accuracies.append(result['final_accuracy'] * 100)
    param_counts.append(result['parameters'])

bars = plt.bar(model_types, accuracies, color=[colors.get(name, 'gray') for name in model_types], alpha=0.7)
plt.title('Final Accuracy Comparison')
plt.ylabel('Test Accuracy (%)')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

# Add parameter count annotations on bars
for bar, param_count in zip(bars, param_counts):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 0.5,
             f'{param_count/1000:.0f}K params', 
             ha='center', va='bottom', fontsize=8)

plt.tight_layout()
plt.show()

# Detailed comparison table
print(f"\n{'='*80}")
print("CNN vs MLP COMPARISON RESULTS")
print(f"{'='*80}")

print(f"\nDetailed Performance Comparison:")
print(f"{'Model':<15}{'Type':<8}{'Parameters':<12}{'Best Acc':<10}{'Final Acc':<11}{'Best Epoch':<12}{'Efficiency':<15}")
print(f"{'-'*80}")

for model_name in ['MLP_Base', 'Simple_CNN', 'Full_CNN']:
    if model_name in results_q17:
        result = results_q17[model_name]
        model_type = result['model_type']
        params = result['parameters']
        best_acc = result['best_accuracy'] * 100
        final_acc = result['final_accuracy'] * 100
        best_epoch = result['best_epoch']
        efficiency = final_acc / (params / 1000)  # Accuracy per 1K parameters
        
        print(f"{model_name:<15}{model_type:<8}{params:<12,}{best_acc:<10.2f}%{final_acc:<11.2f}%{best_epoch:<12}{efficiency:<15.2f}")

# Analysis and insights
print(f"\n{'='*80}")
print("ANALYSIS: CNN vs MLP FINDINGS")
print(f"{'='*80}")

mlp_result = results_q17['MLP_Base']
simple_cnn_result = results_q17['Simple_CNN']
full_cnn_result = results_q17['Full_CNN']

print(f"\n1. PARAMETER EFFICIENCY:")
print(f"   • MLP Base: {mlp_result['parameters']:,} parameters → {mlp_result['final_accuracy']*100:.2f}% accuracy")
print(f"   • Simple CNN: {simple_cnn_result['parameters']:,} parameters → {simple_cnn_result['final_accuracy']*100:.2f}% accuracy")
print(f"   • Full CNN: {full_cnn_result['parameters']:,} parameters → {full_cnn_result['final_accuracy']*100:.2f}% accuracy")

print(f"\n2. CONVERGENCE ANALYSIS:")
print(f"   • MLP Base: Best accuracy at epoch {mlp_result['best_epoch']}")
print(f"   • Simple CNN: Best accuracy at epoch {simple_cnn_result['best_epoch']}")
print(f"   • Full CNN: Best accuracy at epoch {full_cnn_result['best_epoch']}")

print(f"\n3. ACCURACY IMPROVEMENTS:")
mlp_acc = mlp_result['final_accuracy'] * 100
simple_cnn_acc = simple_cnn_result['final_accuracy'] * 100
full_cnn_acc = full_cnn_result['final_accuracy'] * 100

print(f"   • Simple CNN vs MLP: {simple_cnn_acc - mlp_acc:+.2f}% improvement")
print(f"   • Full CNN vs MLP: {full_cnn_acc - mlp_acc:+.2f}% improvement")
print(f"   • Full CNN vs Simple CNN: {full_cnn_acc - simple_cnn_acc:+.2f}% improvement")

print(f"\n4. PARAMETER EFFICIENCY RATIOS:")
mlp_eff = mlp_acc / (mlp_result['parameters'] / 1000)
simple_cnn_eff = simple_cnn_acc / (simple_cnn_result['parameters'] / 1000)
full_cnn_eff = full_cnn_acc / (full_cnn_result['parameters'] / 1000)

print(f"   • MLP Base: {mlp_eff:.2f} accuracy% per 1K parameters")
print(f"   • Simple CNN: {simple_cnn_eff:.2f} accuracy% per 1K parameters")
print(f"   • Full CNN: {full_cnn_eff:.2f} accuracy% per 1K parameters")

print(f"\n{'='*80}")
print("KEY INSIGHTS AND CONCLUSIONS:")
print(f"{'='*80}")

print(f"""
🔍 SPATIAL STRUCTURE ADVANTAGE:
   CNNs leverage the 2D spatial structure of images through:
   • Convolutional layers that detect local features (edges, patterns)
   • Translation invariance - same filter detects features anywhere in image
   • Hierarchical feature learning - simple to complex features

🔍 PARAMETER EFFICIENCY:
   CNNs typically achieve better accuracy with fewer parameters because:
   • Weight sharing in convolutional layers reduces parameter count
   • Pooling layers reduce spatial dimensions while preserving important features
   • Specialized architecture for image data vs generic MLP approach

🔍 CONVERGENCE CHARACTERISTICS:
   • CNNs often converge faster due to inductive biases suited for image data
   • Better generalization through spatial understanding
   • Dropout and architectural choices help prevent overfitting

🔍 PERFORMANCE TRADE-OFFS:
   • Simple CNN: Good balance of parameters vs performance
   • Full CNN: Better accuracy but more parameters (still efficient vs MLP)
   • MLP: Requires many parameters to achieve similar performance

🔍 PRACTICAL IMPLICATIONS:
   • For image classification tasks, CNNs are generally preferred
   • CNN architectures scale better with image complexity
   • Transfer learning opportunities better with CNN features
   • Computational efficiency in practice (despite more operations per parameter)

The results demonstrate that CNNs are fundamentally better suited for image classification
tasks due to their ability to capture spatial relationships and hierarchical features
that MLPs cannot efficiently represent.
""")

# Final summary metrics
best_model = max(results_q17.items(), key=lambda x: x[1]['final_accuracy'])
most_efficient = max(results_q17.items(), key=lambda x: x[1]['final_accuracy'] * 100 / (x[1]['parameters'] / 1000))

print(f"\n📊 SUMMARY METRICS:")
print(f"   • Best Overall Accuracy: {best_model[0]} ({best_model[1]['final_accuracy']*100:.2f}%)")
print(f"   • Most Parameter Efficient: {most_efficient[0]} ({most_efficient[1]['final_accuracy']*100 / (most_efficient[1]['parameters'] / 1000):.2f} acc%/1K params)")
print(f"   • Smallest Model: {min(results_q17.items(), key=lambda x: x[1]['parameters'])[0]} ({min(results_q17.items(), key=lambda x: x[1]['parameters'])[1]['parameters']:,} params)")
print(f"   • Fastest Convergence: {min(results_q17.items(), key=lambda x: x[1]['best_epoch'])[0]} (epoch {min(results_q17.items(), key=lambda x: x[1]['best_epoch'])[1]['best_epoch']})")
````

## Q2.1: Activation Function Analysis (8 points)

In this section, I will experiment with different activation functions (ReLU, Tanh, Sigmoid) and analyze their impact on training performance. I will use identical network architectures and training parameters to ensure fair comparison between activation functions.


````python
# Define networks with different activation functions
class NetworkWithActivation(nn.Module):
    def __init__(self, activation_func, use_xavier=True):
        super(NetworkWithActivation, self).__init__()
        self.flatten = nn.Flatten()
        self.activation_name = activation_func.__name__ if hasattr(activation_func, '__name__') else str(activation_func)
        
        # Define layers
        self.linear1 = nn.Linear(28*28, 512)
        self.activation1 = activation_func()
        self.linear2 = nn.Linear(512, 512) 
        self.activation2 = activation_func()
        self.linear3 = nn.Linear(512, 10)
        
        # Apply Xavier initialization if requested
        if use_xavier:
            self.apply(self._xavier_init)
    
    def _xavier_init(self, module):
        if isinstance(module, nn.Linear):
            nn.init.xavier_uniform_(module.weight)
            nn.init.constant_(module.bias, 0)
    
    def forward(self, x):
        x = self.flatten(x)
        x = self.activation1(self.linear1(x))
        x = self.activation2(self.linear2(x))
        logits = self.linear3(x)
        return logits

# Enhanced training function with detailed tracking
def train_activation_experiment(dataloader, model, loss_fn, optimizer, epoch):
    size = len(dataloader.dataset)
    model.train()
    total_loss = 0
    correct = 0
    num_batches = len(dataloader)
    
    for batch, (X, y) in enumerate(dataloader):
        X, y = X.to(device), y.to(device)
        
        # Compute prediction error
        pred = model(X)
        loss = loss_fn(pred, y)
        total_loss += loss.item()
        
        # Calculate training accuracy
        correct += (pred.argmax(1) == y).type(torch.float).sum().item()
        
        # Backpropagation
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        if batch % 100 == 0:
            loss_val, current = loss.item(), batch * len(X)
            print(f"Epoch {epoch}: loss: {loss_val:>7f}  [{current:>5d}/{size:>5d}]")
    
    avg_loss = total_loss / num_batches
    accuracy = correct / size
    return avg_loss, accuracy

def test_activation_experiment(dataloader, model, loss_fn):
    size = len(dataloader.dataset)
    num_batches = len(dataloader)
    model.eval()
    test_loss, correct = 0, 0
    
    with torch.no_grad():
        for X, y in dataloader:
            X, y = X.to(device), y.to(device)
            pred = model(X)
            test_loss += loss_fn(pred, y).item()
            correct += (pred.argmax(1) == y).type(torch.float).sum().item()
    
    test_loss /= num_batches
    accuracy = correct / size
    return test_loss, accuracy

# Gradient tracking function for specific layer analysis
def get_layer_gradient_norm(model, layer_name='linear2'):
    """Get gradient norm for a specific layer"""
    for name, param in model.named_parameters():
        if layer_name in name and 'weight' in name and param.grad is not None:
            return param.grad.data.norm(2).item()
    return 0.0
````

````python
# Main activation function experiments
activation_functions = {
    'ReLU': nn.ReLU,
    'Tanh': nn.Tanh, 
    'Sigmoid': nn.Sigmoid
}

# Training configuration
epochs = 25
learning_rate = 0.01
results_q2 = {}

print("Starting Activation Function Experiments...")
print("="*70)

# Experiment 1: Different activation functions with Xavier initialization
for act_name, act_func in activation_functions.items():
    print(f"\n{'-'*50}")
    print(f"Training with {act_name} activation (Xavier Init)")
    print(f"{'-'*50}")
    
    # Create model
    model = NetworkWithActivation(act_func, use_xavier=True).to(device)
    
    # Initialize optimizer and loss function
    optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)
    loss_fn = nn.CrossEntropyLoss()
    
    # Track metrics
    train_losses = []
    test_losses = []
    train_accuracies = []
    test_accuracies = []
    gradient_norms = []
    
    # Training loop
    for epoch in range(1, epochs + 1):
        # Train
        train_loss, train_acc = train_activation_experiment(train_dataloader, model, loss_fn, optimizer, epoch)
        
        # Test
        test_loss, test_acc = test_activation_experiment(test_dataloader, model, loss_fn)
        
        # Track gradient norm for middle layer (linear2)
        grad_norm = get_layer_gradient_norm(model, 'linear2')
        
        # Store results
        train_losses.append(train_loss)
        test_losses.append(test_loss)
        train_accuracies.append(train_acc)
        test_accuracies.append(test_acc)
        gradient_norms.append(grad_norm)
        
        if epoch % 5 == 0:
            print(f"Epoch {epoch}: Train Acc: {train_acc:.4f}, Test Acc: {test_acc:.4f}, "
                  f"Train Loss: {train_loss:.4f}, Test Loss: {test_loss:.4f}, Grad Norm: {grad_norm:.4f}")
    
    # Store final results
    results_q2[f"{act_name}_Xavier"] = {
        'model': model,
        'train_losses': train_losses,
        'test_losses': test_losses,
        'train_accuracies': train_accuracies,
        'test_accuracies': test_accuracies,
        'gradient_norms': gradient_norms,
        'final_train_acc': train_accuracies[-1],
        'final_test_acc': test_accuracies[-1],
        'final_train_loss': train_losses[-1], 
        'final_test_loss': test_losses[-1],
        'activation': act_name,
        'initialization': 'Xavier'
    }
    
    print(f"\n{act_name} (Xavier) Final Results:")
    print(f"Train Accuracy: {train_accuracies[-1]:.4f}")
    print(f"Test Accuracy: {test_accuracies[-1]:.4f}")
    print(f"Final Gradient Norm: {gradient_norms[-1]:.4f}")

print(f"\n{'='*70}")
print("Activation Function Experiments Complete!")
print(f"{'='*70}")
````

````python
# CORRECTED: Comprehensive Model Saving for All Experiments (Q1.5, Q1.7, Q2)
import os
import torch

print("="*80)
print("COMPREHENSIVE MODEL SAVING - ALL EXPERIMENTS (Q1.5, Q1.7, Q2)")
print("="*80)

# Create models directory if it doesn't exist
models_dir = "models"
if not os.path.exists(models_dir):
    os.makedirs(models_dir)
    print(f"✓ Created directory: {models_dir}")
else:
    print(f"✓ Directory {models_dir} already exists")

# Dictionary to store all models for saving
all_models_to_save = {}

# Q1.5 Models (Architecture Comparison) - from models_config and results_q15
try:
    print("\n📋 Processing Q1.5 models...")
    for model_name in ['Base', 'Wider', 'Deeper']:
        if model_name in models_config and model_name in results_q15:
            model_key = f"q15_{model_name.lower()}_network"
            all_models_to_save[model_key] = {
                'model': models_config[model_name],
                'results': results_q15[model_name],
                'description': f'Q1.5 {model_name} Network Architecture',
                'experiment': 'Q1.5 Architecture Comparison (Wider vs Deeper)',
                'architecture': 'MLP',
                'activation': 'ReLU',
                'initialization': 'Default PyTorch'
            }
            print(f"   ✓ Added {model_name} Network")
    print(f"✓ Q1.5 models processed: {len([k for k in all_models_to_save.keys() if k.startswith('q15')])} models")
except Exception as e:
    print(f"⚠ Q1.5 models error: {e}")

# Q1.7 Models (CNN vs MLP Comparison) - from all_models and results_q17
try:
    print("\n📋 Processing Q1.7 models...")
    for model_name in results_q17.keys():
        if model_name in all_models:
            model_key = f"q17_{model_name.lower().replace('_', '')}"
            
            # Determine architecture type
            arch_type = 'CNN' if 'CNN' in model_name else 'MLP'
            
            all_models_to_save[model_key] = {
                'model': all_models[model_name],
                'results': results_q17[model_name],
                'description': f'Q1.7 {model_name} Architecture',
                'experiment': 'Q1.7 CNN vs MLP Comparison',
                'architecture': arch_type,
                'activation': 'ReLU',
                'initialization': 'Default PyTorch'
            }
            print(f"   ✓ Added {model_name}")
    print(f"✓ Q1.7 models processed: {len([k for k in all_models_to_save.keys() if k.startswith('q17')])} models")
except Exception as e:
    print(f"⚠ Q1.7 models error: {e}")

# Q2 Models (Activation Function Experiments) - from results_q2
try:
    print("\n📋 Processing Q2 models...")
    for exp_name, results in results_q2.items():
        model_key = f"q2_{exp_name.lower()}"
        all_models_to_save[model_key] = {
            'model': results['model'],
            'results': results,
            'description': f"Q2 {results['activation']} with {results['initialization']} initialization",
            'experiment': 'Q2 Activation Function Analysis',
            'architecture': 'MLP',
            'activation': results['activation'],
            'initialization': results['initialization']
        }
        print(f"   ✓ Added {results['activation']} ({results['initialization']} init)")
    print(f"✓ Q2 models processed: {len([k for k in all_models_to_save.keys() if k.startswith('q2')])} models")
except Exception as e:
    print(f"⚠ Q2 models error: {e}")

# Save all models
print(f"\n💾 Saving {len(all_models_to_save)} models...")
print("-" * 60)

saved_count = 0
failed_count = 0

for model_name, model_data in all_models_to_save.items():
    try:
        save_path = os.path.join(models_dir, f"{model_name}.pth")
        
        # Get model and results
        model = model_data['model']
        results = model_data.get('results', {})
        
        # Prepare base save data
        save_data = {
            'model_state_dict': model.state_dict(),
            'model_architecture': model.__class__.__name__,
            'description': model_data['description'],
            'experiment': model_data['experiment'],
            'architecture_type': model_data['architecture'],
            'activation_function': model_data['activation'],
            'initialization_method': model_data['initialization'],
            'total_parameters': sum(p.numel() for p in model.parameters()),
            'trainable_parameters': sum(p.numel() for p in model.parameters() if p.requires_grad)
        }
        
        # Add experiment-specific data
        if model_name.startswith('q15'):
            # Q1.5 specific data
            save_data.update({
                'final_accuracy': results.get('final_accuracy', 0.0),
                'best_epoch': results.get('best_epoch', 0),
                'final_loss': results.get('final_loss', 0.0),
                'parameters_count': results.get('parameters', 0),
                'training_history': {
                    'train_losses': results.get('train_losses', []),
                    'test_losses': results.get('test_losses', []),
                    'train_accuracies': results.get('train_accuracies', []),
                    'test_accuracies': results.get('test_accuracies', [])
                }
            })
        
        elif model_name.startswith('q17'):
            # Q1.7 specific data
            save_data.update({
                'final_accuracy': results.get('final_accuracy', 0.0),
                'best_epoch': results.get('best_epoch', 0),
                'final_loss': results.get('final_loss', 0.0),
                'parameters_count': results.get('parameters', 0),
                'convergence_epoch': results.get('convergence_epoch', 0),
                'training_history': {
                    'train_losses': results.get('train_losses', []),
                    'test_losses': results.get('test_losses', []),
                    'train_accuracies': results.get('train_accuracies', []),
                    'test_accuracies': results.get('test_accuracies', [])
                }
            })
        
        elif model_name.startswith('q2'):
            # Q2 specific data
            save_data.update({
                'final_train_accuracy': results.get('final_train_acc', 0.0),
                'final_test_accuracy': results.get('final_test_acc', 0.0),
                'final_train_loss': results.get('final_train_loss', 0.0),
                'final_test_loss': results.get('final_test_loss', 0.0),
                'final_gradient_norm': results.get('gradient_norms', [0.0])[-1],
                'training_history': {
                    'train_losses': results.get('train_losses', []),
                    'test_losses': results.get('test_losses', []),
                    'train_accuracies': results.get('train_accuracies', []),
                    'test_accuracies': results.get('test_accuracies', []),
                    'gradient_norms': results.get('gradient_norms', [])
                }
            })
        
        # Save the model
        torch.save(save_data, save_path)
        print(f"✓ Saved: {model_name:<25} -> {model_data['description']}")
        saved_count += 1
        
    except Exception as e:
        print(f"✗ Failed to save {model_name}: {str(e)}")
        failed_count += 1

# Create comprehensive summary file
summary_path = os.path.join(models_dir, "comprehensive_model_summary.txt")
with open(summary_path, 'w') as f:
    f.write("COMPUTER VISION ASSIGNMENT 3 - COMPREHENSIVE MODEL SUMMARY\n")
    f.write("=" * 70 + "\n\n")
    f.write(f"Total Models Saved: {saved_count}\n")
    f.write(f"Failed Saves: {failed_count}\n")
    f.write(f"Save Date: {torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU'}\n\n")
    
    # Q1.5 Models Summary
    f.write("QUESTION 1.5 MODELS (Architecture Comparison):\n")
    f.write("-" * 50 + "\n")
    for model_name, model_data in all_models_to_save.items():
        if model_name.startswith('q15'):
            results = model_data.get('results', {})
            f.write(f"{model_name}: {model_data['description']}\n")
            f.write(f"  Parameters: {results.get('parameters', 'Unknown'):,}\n")
            f.write(f"  Final Accuracy: {results.get('final_accuracy', 0.0):.4f}\n")
            f.write(f"  Architecture: {model_data['architecture']}\n\n")
    
    # Q1.7 Models Summary
    f.write("QUESTION 1.7 MODELS (CNN vs MLP Comparison):\n")
    f.write("-" * 50 + "\n")
    for model_name, model_data in all_models_to_save.items():
        if model_name.startswith('q17'):
            results = model_data.get('results', {})
            f.write(f"{model_name}: {model_data['description']}\n")
            f.write(f"  Parameters: {results.get('parameters', 'Unknown'):,}\n")
            f.write(f"  Final Accuracy: {results.get('final_accuracy', 0.0):.4f}\n")
            f.write(f"  Architecture: {model_data['architecture']}\n")
            f.write(f"  Best Epoch: {results.get('best_epoch', 'Unknown')}\n\n")
    
    # Q2 Models Summary
    f.write("QUESTION 2 MODELS (Activation Function Analysis):\n")
    f.write("-" * 50 + "\n")
    for model_name, model_data in all_models_to_save.items():
        if model_name.startswith('q2'):
            results = model_data.get('results', {})
            f.write(f"{model_name}: {model_data['description']}\n")
            f.write(f"  Test Accuracy: {results.get('final_test_acc', 0.0):.4f}\n")
            f.write(f"  Final Loss: {results.get('final_test_loss', 0.0):.4f}\n")
            f.write(f"  Parameters: {sum(p.numel() for p in model_data['model'].parameters()):,}\n")
            f.write(f"  Activation: {model_data['activation']}\n")
            f.write(f"  Initialization: {model_data['initialization']}\n\n")

print(f"\n" + "="*60)
print(f"COMPREHENSIVE MODEL SAVING COMPLETE!")
print(f"✓ Successfully saved: {saved_count} models")
print(f"✗ Failed to save: {failed_count} models")
print(f"✓ Models directory: {os.path.abspath(models_dir)}")
print(f"✓ Summary file: {os.path.abspath(summary_path)}")
print("="*60)

# Display directory contents with model categories
print(f"\nSaved models in '{models_dir}':")
print("📁 MODEL INVENTORY:")

try:
    q15_files = [f for f in os.listdir(models_dir) if f.startswith('q15')]
    q17_files = [f for f in os.listdir(models_dir) if f.startswith('q17')]
    q2_files = [f for f in os.listdir(models_dir) if f.startswith('q2')]
    other_files = [f for f in os.listdir(models_dir) if not any(f.startswith(x) for x in ['q15', 'q17', 'q2'])]
    
    if q15_files:
        print(f"\n  🏗️  Q1.5 Architecture Models ({len(q15_files)}):")
        for file in sorted(q15_files):
            size_mb = os.path.getsize(os.path.join(models_dir, file)) / (1024*1024)
            print(f"     • {file:<35} ({size_mb:.2f} MB)")
    
    if q17_files:
        print(f"\n  🧠 Q1.7 CNN vs MLP Models ({len(q17_files)}):")
        for file in sorted(q17_files):
            size_mb = os.path.getsize(os.path.join(models_dir, file)) / (1024*1024)
            print(f"     • {file:<35} ({size_mb:.2f} MB)")
    
    if q2_files:
        print(f"\n  ⚡ Q2 Activation Models ({len(q2_files)}):")
        for file in sorted(q2_files):
            size_mb = os.path.getsize(os.path.join(models_dir, file)) / (1024*1024)
            print(f"     • {file:<35} ({size_mb:.2f} MB)")
    
    if other_files:
        print(f"\n  📋 Other Files ({len(other_files)}):")
        for file in sorted(other_files):
            file_path = os.path.join(models_dir, file)
            if os.path.isfile(file_path):
                size_mb = os.path.getsize(file_path) / (1024*1024)
                print(f"     • {file:<35} ({size_mb:.2f} MB)")
    
    total_models = len(q15_files) + len(q17_files) + len(q2_files)
    print(f"\n📊 TOTAL: {total_models} trained models saved")
    
except Exception as e:
    print(f"  Could not list directory contents: {e}")

print(f"\n🎉 All models from Q1.5, Q1.7, and Q2 are now saved!")
print(f"🚀 Ready for future applications, analysis, and deployment!")
````
